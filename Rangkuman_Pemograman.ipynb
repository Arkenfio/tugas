# *Rangkuman Pemograman*
Pemrograman adalah proses membuat instruksi untuk komputer dengan menggunakan bahasa pemrograman seperti Python, C++, atau Java. Konsep dasar dalam pemrograman meliputi:

1. **Variabel**: Tempat menyimpan data, misalnya `x = 5` atau `nama = 'Andi'`.
2. **Tipe Data**: Jenis data yang bisa disimpan, seperti Integer, Float, String, dan Boolean.
3. **Operator**: Digunakan untuk operasi pada variabel, seperti operator aritmatika (`+`, `-`), perbandingan (`==`, `!=`), dan logika (`and`, `or`).
4. **Struktur Kendali**:
   - **Pengkondisian (If-Else)**: Mengatur alur program berdasarkan kondisi, misalnya `if nilai >= 75`.
   - **Perulangan (Loop)**: Mengulangi blok kode beberapa kali, misalnya `for` dan `while`.
5. **Logika Pemrograman**: Urutan langkah sistematis untuk menyelesaikan masalah.
6. **Struktur Algoritma Dasar**: Langkah-langkah seperti memasukkan angka, melakukan operasi, dan mencetak hasil.

Pemahaman dasar ini penting untuk membuat program yang efisien dan terstruktur.
